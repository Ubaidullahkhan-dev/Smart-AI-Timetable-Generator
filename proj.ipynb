# Create Basic Timetable Structure 

```python
DAYS = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

TIME_SLOTS = [
    "9-10",
    "10-11",
    "11-12",
    "1-2",
    "2-3"
]

SUBJECTS = [
    "Math",
    "Physics",
    "Programming",
    "English"
]

TEACHERS = {
    "Math": "Ali",
    "Physics": "Ahmed",
    "Programming": "Sara",
    "English": "Ayesha"
}
ROOMS = ["R1", "R2", "Lab1"]
```

# Create Random Timetable

```python
import random


def create_random_timetable():
    timetable = {}

    for day in DAYS:
        timetable[day] = {}

        for slot in TIME_SLOTS:
            subject = random.choice(SUBJECTS)#ramdomly select subject
            teacher = TEACHERS[subject]#select teacher related to sub
            room = random.choice(ROOMS)#select room randomly

            timetable[day][slot] = {
                "subject": subject,
                "teacher": teacher,
                "room": room
            }

    return timetable
```

# Print Timetable

```python
def print_timetable(timetable):
    for day, slots in timetable.items():
        print(f"\n{day}")

        for slot, details in slots.items():
            print(
                slot,
                details["subject"],
                details["teacher"],
                details["room"]
            )
```

# Clash Detection

```python

def calculate_clashes(timetable):
    clashes = 0   
    for slot in TIME_SLOTS:
        teachers_used = []
        rooms_used = []

        for day in DAYS:
            lecture = timetable[day][slot]

            teacher = lecture["teacher"]
            room = lecture["room"]

            if teacher in teachers_used:
                clashes += 1
            else:
                teachers_used.append(teacher)

            if room in rooms_used:
                clashes += 1
            else:
                rooms_used.append(room)

    return clashes
```

# Fitness Function

```python
def fitness(timetable):
    clashes = calculate_clashes(timetable)

    return 100 - (clashes * 10)
```

# Generate Population

```python
def create_population(size):
    population = []

    for i in range(size):
        timetable = create_random_timetable()
        population.append(timetable)

    return population
```

# Selection

```python
def selection(population):
    population.sort(key=fitness, reverse=True)

    return population[:2]
```

# Crossover

```python
def crossover(parent1, parent2):
    child = {}

    for day in DAYS:
        if random.random() > 0.5:
            child[day] = parent1[day]
        else:
            child[day] = parent2[day]

    return child
```

# Mutation

```python
def mutate(timetable):
    day = random.choice(DAYS)
    slot = random.choice(TIME_SLOTS)

    new_subject = random.choice(SUBJECTS)

    timetable[day][slot]["subject"] = new_subject
    timetable[day][slot]["teacher"] = TEACHERS[new_subject]

    return timetable
```

# Genetic Algorithm Main Engine

```python

def genetic_algorithm(generations=100, population_size=10):
    population = create_population(population_size)

    for generation in range(generations):
        parents = selection(population)

        new_population = []

        for _ in range(population_size):
            child = crossover(parents[0], parents[1])

            child = mutate(child)

            new_population.append(child)

        population = new_population

    best = selection(population)[0]

    return best
```

# Run Project

```python
best_timetable = genetic_algorithm()

print_timetable(best_timetable)

print("Fitness:", fitness(best_timetable))
```
